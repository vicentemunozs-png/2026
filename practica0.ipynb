#Control de flujos

```python
#Condicionales
venta_cerezas = 50000

if venta_cerezas > 200000:
  categoria = "Excelente"
  print("Se guardó categoría")
elif venta_cerezas > 100000:      #va un elif, despues del if para seguir otorgando condiciones
  categoria = "Buena"
  print("Se guardó categoría")
elif venta_cerezas >75000:
  categoria = "Normal"
  print("No sé si se guardó")
else:
  categoria = "Mala"              #un else debido a que ya no sigo otorgando condiciones y el resto va a caer acá
  print(categoria)
```

```python
#bucle: for
ventas_frutas = [15000,25000,36000,27890,14700]
for v in ventas_frutas:
  print("Total ventas", v)
  print("IVA_transacción", 0.19*v)
  print()

print("Se terminó el iterable")
```
